```python
import random

def montecarlo_pi(number_of_shots): 
    number_of_hits = 0
    
    for i in range(number_of_shots):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
        if x*x + y*y < 1:
            number_of_hits += 1
            
    return 4*number_of_hits/number_of_shots
```

```python
montecarlo_pi(50)
```

```python
import numpy
import math
import matplotlib.pyplot as plt

absolute_error = []
relative_error = []

for i in range(0,100):
    calculated_pi = montecarlo_pi(1000)
    true_pi = numpy.pi
    abs_error = abs(true_pi - calculated_pi)
    rel_error = abs((true_pi - calculated_pi)/(true_pi))
    
    absolute_error.append(abs_error)
    relative_error.append(rel_error)

    print(calculated_pi, true_pi)
```

```python
plt.figure()
plt.plot(absolute_error, marker= 'o')
plt.xlabel("Run Number")
plt.ylabel("Absolute Error")
plt.title("Absolute Error of Monte Carlo pi")
plt.show()

plt.figure()
plt.plot(relative_error, marker= 'o')
plt.xlabel("Run Number")
plt.ylabel("Relative Error")
plt.title("Relative Error of Monte Carlo pi")
plt.show()
```
