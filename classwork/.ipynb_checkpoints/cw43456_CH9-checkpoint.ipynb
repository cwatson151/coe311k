Part A

```python
import numpy as np

# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

print("X:")
print(X)

print("Y:")
print(Y)

print("A:")
print(A)
```

```python
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
```

# Plot eah data point
# Plot the line y = 0.95*x + 59

```python
import matplotlib.pyplot as plt

Y_line = m*X + c

plt.scatter(X, Y, label="Data points")
plt.plot(X, Y_line, label="Best fit line")

plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```

Randomize the vectors

```python
import numpy as np
perm = np.random.permutation(len(X))
X_shuffled = X[perm]
Y_shuffled = Y[perm]

print(X_shuffled)
print(Y_shuffled)
```

Part B

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.arange(0, 1.01, 0.01)
X = np.sort(X)
m = np.random.uniform(-5, 5)
b = np.random.uniform(-2, 2)

error = np.random.uniform(0, 1, size=len(X))
sign = np.random.choice([-1, 1], size=len(X))
random_error = sign * error

Y = m * X + b + random_error
print(f"Line: y = {m:.6f}x + {b:.6f}")
```

```python
# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T
print("A:")
print(A)

# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Best fit: y = {m:.6f}x + {c:.6f}")
```

```python
#Residuals and extremes
Y_predicted = m * X + c
residuals = Y - Y_predicted

print("\nResiduals (first 10):")
print(residuals[:10])

#Largest and Smallest Residuals
idx_max = np.argmax(residuals)
idx_min = np.argmin(residuals)

print("\nMost positive residual:")
print(f"  index={idx_max}, X={X[idx_max]:.2f}, Y={Y[idx_max]:.6f}, Y_predicted={Y_predicted[idx_max]:.6f}, residual={residuals[idx_max]:.6f}")

print("\nMost negative residual:")
print(f"  index={idx_min}, X={X[idx_min]:.2f}, Y={Y[idx_min]:.6f}, Y_predicted={Y_predicted[idx_min]:.6f}, residual={residuals[idx_min]:.6f}")
```

```python
#Plot Points and Predicted line
plt.scatter(X, Y, label="Data points")
plt.plot(X, Y_predicted, label="Best fit line")

plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```
