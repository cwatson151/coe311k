```python
import numpy as np
```

```python
v1 = np.array([1, 2, 3]) #3D vector
v2 = np.array([4, 5, 6])
```

```python
print (v_add)
print (v_sub)
```

```python
print (dot_product)
```

```python
A = np.array([[1, 2, 3], [4, 5, 6]])
print(A)
print()

B = np.array([[7, 8, 9], [10, 11, 12], [13, 14, 15]])
print(B)
```

```python
C = np.array([[1, 2, 3], [4, 5, 6]])
matrix_add = A + C
matrix_sub = A - C

print("A+C=\n", matrix_add)
print("A-C=\n", matrix_sub)
```

```python
matrix_mul = np.dot(A, B)

print(matrix_mul)
```

```python
transpose_A = A.T
print(transpose_A)
```

```python
A = np.array([[2, 3], [4, -1]])
b = np.array([5, 3])
print(A)
print(b)

x = np.linalg.solve(A, b)
print ("Solution:")
print(x)
```

```python
A = np.array([[2, 1], [1, 2]])
e_value, e_vector = np.linalg.eig(A)
print ("eiganvalue:", e_value)
print("eiganvector:", e_vector)
```

```python
A = np.array([[4, 7], [2, 6]])
A_inv = np.linalg.inv(A)

print(A)
print(A_inv)
```

```python
identity = np.dot(A, A_inv)
print(identity)
```

```python
# Define a matrix
A = np.array([[1, 2], [3, 4], [5, 6]])

# Perform SVD
U, sigma, Vt = np.linalg.svd(A)

print("U matrix:\n", U)
print("Sigma values:", sigma)
print("V^T matrix:\n", Vt)
```

Gaussian Elimination

```python
import numpy as np
def gaussian_elimination(A, b):
    n = len(b)
    # Forward elimination
    for i in range(n):
        # Pivoting
        max_row = max(range(i, n), key=lambda x: abs(A[x][i]))
        A[i], A[max_row] = A[max_row], A[i]
        b[i], b[max_row] = b[max_row], b[i]

        # Make the rows below this one 0 in current column
        for j in range(i + 1, n):
            ratio = A[j][i] / A[i][i]
            A[j] = [A[j][k] - ratio * A[i][k] for k in range(n)]
            b[j] -= ratio * b[i]
# Backward substitution
    x = [0] * n
    for i in range(n - 1, -1, -1):
        x[i] = (b[i] - sum(A[i][j] * x[j] for j in range(i + 1, n))) / A[i][i]
    return x
```

```python
# Example Usage
A = [[3, 2, -4], [2, 3, 3], [5, -3, 1]]
b = [3, 15, 14]
solution = gaussian_elimination(A, b)
print(f"Solution: {solution}")
```

```python
ans = np.linalg.solve(A,b)
print(ans)
```
