```python
import numpy as np
import matplotlib.pyplot as plt

# RHS of differential equation
def f(t, y):
    return -1000*y + 3000 - 2000*np.exp(-t)

# Solution
def y_exact(t):
    return 3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t)

# Euler Forward
def euler_forward(f, t0, y0, tf, h):
    n = int((tf - t0) / h)
    t = np.zeros(n + 1)
    y = np.zeros(n + 1)

    t[0] = t0
    y[0] = y0

    for i in range(n):
        t[i+1] = t[i] + h
        y[i+1] = y[i] + h * f(t[i], y[i])

    return t, y

# Parameters
t0 = 0
tf = 30
y0 = 0
h_values = [0.1, 0.01, 0.001, 0.00001, 0.0000001]

# Exact solution for plotting
t_exact = np.linspace(t0, tf, 5000)
y_true = y_exact(t_exact)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(t_exact, y_true, label='Analytical Solution', linewidth=2)

for h in h_values:
    t_num, y_num = euler_forward(f, t0, y0, tf, h)
    plt.plot(t_num, y_num, label=f"Euler Forward, h={h}")

plt.xlabel("t")
plt.ylabel("y(t)")
plt.title("Euler Forward vs Analytical Solution")
plt.legend()
plt.grid(True)
plt.show()

# Final values at t = 30
print("Final values at t = 30:\n")
print(f"Analytical: y(30) = {y_exact(30):.10f}")

for h in h_values:
    t_num, y_num = euler_forward(f, t0, y0, tf, h)
    print(f"Euler Forward (h={h}): y(30) = {y_num[-1]:.10f}")
```
