```python
import numpy as np

A = np.array([
	[0.582745, 0.48	, 0.10	, 0.  	, 0.  	], 
 [0.48	, 1.044129, 0.46	, 0.10	, 0.  	],
 [0.10	, 0.46	, 1.10431 , 0.44	, 0.10	],
 [0.  	, 0.10	, 0.44	, 0.963889, 0.42	],
 [0.  	, 0.  	, 0.10	, 0.42	, 0.522565]], dtype=float)

def diagonally_dominant(matrix):
    n = len(matrix)
    for i in range(n):
        if abs(matrix[i, i]) < sum(abs(matrix[i, j]) for j in range(n) if j != i):
            return False
    return True
print("Is the matrix diagonally dominant?", diagonally_dominant(A))
```
