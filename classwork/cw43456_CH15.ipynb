```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -2 * y

# Exact solution
def y_exact(x):
    return np.exp(-2 * x)

# Define parameters
x_start, x_end = 0, 2

# Step sizes
h_values = np.arange(0.01, 1.01, 0.05)

first_x = None
first_y = None

# Store total accumulated errors
total_errors = []

# Create figure for Euler runs
plt.figure(figsize=(10, 6))

# Loop over step sizes
for run_idx, h in enumerate(h_values):
    num_steps = int((x_end - x_start) / h)
    x_values = x_start + np.arange(num_steps + 1) * h
    y_values = np.zeros(num_steps + 1)
    
    # Initial Condition
    y_values[0] = 1
    
    # Euler's Method
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
         # Plot
    plt.plot(x_values, y_values, label=f"h = {h:.2f}")
    
    # First run
    if run_idx == 0:
        first_x = x_values.copy()
        first_y = y_values.copy()
        total_error = 0
        
        print(f"\nRun {run_idx+1}, h = {h:.2f} (reference)")
        for i in range(len(x_values)):
            print(f"x = {x_values[i]:.4f}, error = 0.000000")
    else:
        # Interpolate current to first run
        y_interpolated = np.interp(first_x, x_values, y_values)
        
        errors = np.abs(y_interpolated - first_y)
        total_error = np.sum(errors)
        
        print(f"\nRun {run_idx+1}, h = {h:.2f}")
        for i in range(len(first_x)):
            print(f"x = {first_x[i]:.4f}, error = {errors[i]:.6f}")
    
    print(f"Total accumulated error = {total_error:.6f}")
    total_errors.append((h, total_error))

# Plot exact solution
x_exact = np.linspace(x_start, x_end, 500)
plt.plot(x_exact, y_exact(x_exact), 'k--', linewidth=2, label='Exact solution')

plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler's Method for dy/dx = -2y with Multiple Step Sizes")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot total accumulated error vs step size
h_plot = [item[0] for item in total_errors]
err_plot = [item[1] for item in total_errors]

plt.figure(figsize=(10, 6))
plt.plot(h_plot, err_plot, marker='o')
plt.xlabel("Step size h")
plt.ylabel("Total accumulated error vs first run")
plt.title("Total Accumulated Error vs Step Size")
plt.grid(True)
plt.show()
```
