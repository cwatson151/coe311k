```python
def convertBase10toBase2(num):
    base10 = ' '
    while num/2 != 0:
        remainder = num % 2
        base10 += str(remainder)
        num = num // 2
        
    base10 = base10[::-1]
    return base10
    print(convertBase10toBase2(20))
    print(convertBase10toBase2(2025))
```

```python
convertBase10toBase2(20)
```

```python
convertBase10toBase2(2025)
```
