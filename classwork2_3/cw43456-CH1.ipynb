```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2*np.pi, 50)
```

```python
x
```

```python
plt.plot(x, np.sin(x))
plt.show()
```

```python
plt.subplot(2, 2, 1) # (row, column, active area)
plt.plot(x, np.sin(x), 'r')
plt.subplot(2, 2, 2)
plt.title("Line")
plt.scatter(x, np.cos(x), color= 'g')
plt.title("Scatter")
plt.subplot(2, 2, 3)
plt.bar(x[:10], (np.e**x)[:10])
plt.title("Bar")
plt.subplot(2, 2, 4)
plt.hist(np.tan(x), bins=20)
plt.title("Histogram")
plt.show()
```
