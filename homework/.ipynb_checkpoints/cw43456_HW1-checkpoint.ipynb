```python
import random
import numpy
import math
import matplotlib.pyplot as plt
```

Monte Carlo Pi Function

```python
def montecarlo_pi(number_of_shots): 
    number_of_hits = 0
    
    for i in range(number_of_shots):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
        if x*x + y*y < 1:
            number_of_hits += 1
            
    return 4*number_of_hits/number_of_shots
```

Main Code

```python
absolute_error = []
steps = []

true_pi = numpy.pi
for i in range(10,1001,20):
    calculated_pi = montecarlo_pi(i)
    abs_error = abs(true_pi - calculated_pi)
    steps.append(i)
    absolute_error.append(abs_error)
```

Plot Routine

```python
plt.figure()
plt.plot(steps, absolute_error)
plt.xlabel("Number of Steps")
plt.ylabel("Absolute Error")
plt.title("Monte Carlo Pi Error")
plt.show
```
