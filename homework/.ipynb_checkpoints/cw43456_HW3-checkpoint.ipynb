```python
import math
import numpy as np
import matplotlib.pyplot as plt

def taylor_sin(x, n_terms):
    x = np.asarray(x, dtype=float)
    term = x.copy()
    psum = term.copy()
    
    for k in range(1, n_terms):
        term *= (-1)*x*x/((2*k)*(2*k+1))
        psum += term
    return psum #partial sum of Taylor series
```

```python
#Compute the approximate values of sin⁡(x) for x=0 to 10, steps of .1 using the truncated series with 1, 2, 3, and 4 terms.
x = np.arange(0, 10.0 + 0.1, 0.1)

#Calculate the exact value of sin⁡(x)
exact_sin = np.sin(x)

#Compare the Taylor series approximations to the exact value and calculate the absolute error
approx_sin = {}
error = {}
for n in [1, 2, 3, 4]:
    approx_sin[n] = taylor_sin(x, n)
    error[n] = np.abs(approx_sin[n] - exact_sin)
```

```python
#Plot truncation error vs x for each term 
plt.figure()
for n in [1, 2, 3, 4]:
    plt.plot(x, error[n], label=f"{n} term(psum)")
plt.xlabel("x")
plt.ylabel("Error")
plt.title("Truncation Error for Taylor Series Approximation of sin(x)")
plt.legend()
plt.grid(True)
plt.show()
```

The truncation error reduces when you add more terms. For small x, the terms decrease quickly because the term x^2k+1 is small. For larger x-values, the powers in the term x^2k+1 increase very quickly and can deviate from the true sin(x). Therefore, the truncation error decreases with more terms, but the rate is more favorable at small x-values than it is at large x-values.
